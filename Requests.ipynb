```python
#Create an API client
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-4-5-20251001"
```

```python
def add_user_msg(msg, text):
    user_message = {"role": "user", "content": text}
    msg.append(user_message)

def add_ai_msg(msg, text):
    ai_msg = {"role": "assistant", "content": text}
    msg.append(ai_msg)

#Make a request
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

Functions to operate multiple chats


```python
#Make a starting list of messages
messages = []

#Add in the initaial user question of "Defin node JS in one sentence"
add_user_msg(messages, "define Node JS in one sentence")

#Pass the lksit of messages into t'chat' to get an answer
answer = chat(messages)


#Take the answer and add it as an assistant message into our list
add_ai_msg(messages, answer)

#Add in the user's follow-up question
add_user_msg(messages, "Write another sentence")

#Call chat again with the list of messages to get a final answer
aswer = chat(messages)
answer
```
